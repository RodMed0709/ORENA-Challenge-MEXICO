# Rung 08 — Data Card

Generates every table and every headline number in `README.md`, which **is** the card.

No GPU, no pod, seconds to run. It reads the 410 KB of QA parquets from `external_data/`
(gitignored — see README §Reproducing) and the rung-02 eval artifacts already in the repo.

**The asserts are the point.** Every figure the card quotes is checked in
`_tools/build_card.py`. If a number stops being true, this notebook raises instead of
letting the card lie. A failing assert is a finding — never move the expected value.

```python
DATA_ROOT = "external_data/orena-data"
EVAL_DIR = "experiments/02-lora-sft/runs/02_lora_sft_v1/eval_best"
```

```python
import sys
from pathlib import Path

import pandas as pd

EXP_DIR = Path.cwd()
REPO = EXP_DIR
while REPO != REPO.parent and not ((REPO / ".git").exists() or (REPO / "src").is_dir()):
    REPO = REPO.parent
for p in (EXP_DIR / "_tools", REPO / "src"):
    if p.is_dir():
        sys.path.insert(0, str(p))

from build_card import build

h = build(
    data_root=REPO / DATA_ROOT,
    eval_dir=REPO / EVAL_DIR,
    out_dir=EXP_DIR / "tables",
)
print("OK — every headline in README.md re-asserted against the data.\n")
for k, v in h.items():
    print(f"  {k:32s} {v}")
```

```python
# The tables the card embeds. Committed: they are the deliverable, not a run artifact.
for f in sorted((EXP_DIR / "tables").glob("*.csv")):
    print(f"\n=== {f.name} ===")
    print(pd.read_csv(f).to_string(index=False, max_colwidth=60))
```

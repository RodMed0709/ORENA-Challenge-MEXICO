````python
# papermill parameters
SMOKE = True   # one frame per group, prints timings so N is chosen from a measurement
N_C1 = 40      # frames per group (gold >= 5, gold == 1)
N_C2 = 30      # adjacent-frame pairs
GRID = 16      # point lattice; GRID**2 model calls per frame
SEED = 42
````

````python
import json
import sys
from pathlib import Path

EXP = Path.cwd()
sys.path.insert(0, str(EXP / "_tools"))

from run_controls import Config, main

cfg = Config(smoke=SMOKE, n_c1=N_C1, n_c2=N_C2, grid=GRID, seed=SEED)
result = main(cfg)
print(json.dumps({"C1": result["C1"], "C2": result["C2"]}, indent=2))
````
